```python
cust_schema = 'customer_id long, purchase_date date, product_id long, amount double'
```

```python
cust_df = spark.read.schema(cust_schema).csv('/public/trendytech/datasets/cust_transf.csv')
```

```python
def get_cust_hist1(customer_id):
    return cust_df.filter(cust_df.customer_id == customer_id)
```

```python
history_1a = get_cust_hist1("1004")
```

```python
history_1a.count()
```

```python
cust_df_repart =cust_df.repartition(200,"customer_id")
```

```python
cust_df_repart.rdd.getNumPartitions()
```

```python
def get_cust_hist(customer_id):
    return cust_df_repart.filter(cust_df.customer_id == customer_id)
```

```python
cust_df_repart.persist(StorageLevel.MEMORY_AND_DISK)
```
