## Cache and Persist

```python
order = 'order_id long , order_date date , cust_id long, status string'
```

```python
df1 = spark.read.option("inferSchema","true").schema(order).option("header","false").csv('/public/trendytech/orders/orders_1gb.csv')
```

```python
df1.count()
```

```python
df1.count()
```

```python
df1_cached = df1.cache()
```

```python
df1_cached.show()
```

```python
df1_cached.head()
```

#### even though we have cached the df. it wont happen immediatly. Only when next action is called it will be done. that too only the data required for the action will be cached

```python
df1_cached.tail(1)
```

```python
df1_cached.head()
```

```python
df1_cached.count()
```

#### now all the data will be cached as the action requires whole data. now subsequent action will be much faster

```python
df1_cached.head()
```

```python
df1_cached.tail(2)
```

```python
df1_cached.count()
```

```python
df1_cached.unpersist() ## remove cache
```
