## Dataframe writer API

#### 4 modes - override, ignore, append, error if exists

```python
orders_df.write \
.mode('overwrite') \
.format('csv') \
.option('path','/user/itv027484/sparkwriter') \
.save()
```

```python
orders_df.write \
.mode('overwrite') \
.option('path','/user/itv027484/sparkwriterPQ') \
.save()  #by default format is parquet
```

```python
orders_df.write \
.mode('overwrite') \
.format('orc') \
.option('path','/user/itv027484/sparkwriterORC') \
.save()
```

```python
orders_df.write \
.mode('append') \
.format('csv') \
.option('path','/user/itv027484/sparkwriter') \
.save()
```

```python
orders_df.write \
.mode('errorifExists') \
.format('csv') \
.option('path','/user/itv027484/sparkwriter') \
.save()
```

```python
orders_df.write \
.mode('ignore') \
.format('orc') \
.option('path','/user/itv027484/sparkwriterORC') \
.save()  ## will not write if target folder exists
```
