#### groupbykey vs reduceByKey

```python
rdd_1gb = spark.sparkContext.textFile("/public/trendytech/orders/orders_1gb.csv")
```

['1,2013-07-25 00:00:00.0,11599,CLOSED',

```python
rdd2 = rdd_1gb.map(lambda x: (x.split(",")[3],1))
```

```python
rdd3 = rdd2.reduceByKey(lambda x,y : x+y)  ## its a wide transformation
```

```python
rdd3.count()
```

```python
rdd_1gb.getNumPartitions()
```

###############

```python
rdd4 = rdd_1gb.map(lambda x: (x.split(",")[3],x.split(",")[2]))
```

```python
rdd5 = rdd4.groupByKey()  ## its also a wide transformation but high chane of out of memory error, as no local aggregation is peformed
```

```python
rdd6 = rdd5.map(lambda x: (x[0],len(x[1])))
```

```python
rdd6.count()
```
