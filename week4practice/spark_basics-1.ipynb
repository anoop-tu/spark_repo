#### RDD - Word Count example

```python
from pyspark.sql import SparkSession
import getpass
username = getpass.getuser()
spark = SparkSession. \
builder. \
config('spark.ui.port','0'). \
config("spark.sql.warehouse.dir", f"/user/itv024128/warehouse"). \
enableHiveSupport(). \
master('yarn'). \
getOrCreate()
```

```python
rdd1 = spark.sparkContext.textFile("/user/itv024128/data.txt")
```

```python
rdd1.take(5)
```

```python
rdd2 = rdd1.flatMap(lambda line: line.split(" "))    
```

```python
rdd2.take(10)
```

```python
rdd3 = rdd2.map(lambda x: (x,1))
```

```python
rdd3.take(5)
```

```python
rdd4 = rdd3.reduceByKey(lambda x,y : x+y)
```

```python
rdd4.take(5)
```

```python
rdd5 = rdd4.sortBy(lambda x:x[1],False)
```

```python
rdd5.take(5)
```

```python
rdd4.saveAsTextFile("/user/itv024128/rdd.txt")  
```
