```python
from pyspark.sql import SparkSession
import getpass
username = getpass.getuser()
spark = SparkSession. \
builder. \
appName("cache rdd"). \
config('spark.ui.port','0'). \
config("spark.sql.warehouse.dir", f"/user/itv024128/warehouse"). \
enableHiveSupport(). \
master('yarn'). \
getOrCreate()
```

```python
orders_rdd = spark.sparkContext.textFile("/public/trendytech/orders/orders_1gb.csv")
```

```python
orders_filtered = orders_rdd.filter(lambda x: x.split(",")[3] != 'PENDING PAYMENT')
```

```python
#orders_filtered.take(5)
```

```python
order_mapped = orders_filtered.map(lambda x: (x.split(",")[2],1))
```

```python
#order_mapped.take(5)
```

```python
order_reduced = order_mapped.reduceByKey(lambda x,y : x+y)
```

```python
#order_reduced.take(5)
```

```python
result = order_reduced.filter(lambda x: int(x[0]) < 501)
```
