### filter() & where()

```python
df_south = retail_sales_df.where("Region = 'South'")
```

```python
df_south.show(3)
```

```python
df_south = retail_sales_df.where(col("Region") == 'South')  ## col - requires from pyspark.sql.functions import col
```

```python
df_south.show(3)
```

```python
df_south = retail_sales_df.where(retail_sales_df.Region == 'South')
```

```python
df_south.show(3)
```

```python
df_south = retail_sales_df.filter(retail_sales_df.Region == 'South')
```

```python
df_south.show(3)
```

```python
df_south = retail_sales_df.filter(col("Region") == 'South')
```

```python
df_south.show(3)
```
