**A decorator in Python is a function that takes another function (or class) as input, wraps extra behavior around it, and returns a new function — without you having to change the original function's code.**

```python
def greet():
    print('hello!')
```

```python
def my_decorator(func):
    def wrapper():
        print('Before func exec...')
        func()
        print('After func exec...')
    return wrapper
```

```python
wrap = my_decorator(greet)
wrap()
```

```python
@my_decorator
def greet():
    print('hello!')
```

```python
greet()
```

```python
def write_to_delta(func):
    def wrapper():
        df = func()
        table_name = 'ext_cat.default.orders'
        df.write.format('delta').saveAsTable(table_name)
        print('Delta table created')
    return wrapper
```

```python
@write_to_delta
def orders_bronzelatest():
    df = (
        spark.read
        .format('csv')
        .option('header','true')
        .option('inferSchema','true')
        .load('/Volumes/ext_cat/default/extvol/orders/orders1.csv')
    )
    return df
```
