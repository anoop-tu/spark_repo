Try to load a new csv in the same folder but with a slightly different layout - some data in columns are of differen data type

```python
%sql
drop table if exists employees;
create table employees as(
select * from 
read_files (
    '/Volumes/ext_cat/default/extvol/emp/',
    format => 'csv',
    header => 'true'
))
```

```python
%sql
select * from employees
```

Department_ID column changed from INT TO STRING as one of the file had a string value in this column - Schema is inferred

Below approach enforces schema

```python
%sql
drop table if exists employees;
create table employees as(
select * from 
read_files (
    '/Volumes/ext_cat/default/extvol/emp/',
    format => 'csv',
    header => 'true',
    schema => 'EMPLOYEE_ID int,FIRST_NAME string,LAST_NAME string,EMAIL string, PHONE_NUMBER string,HIRE_DATE string, JOB_ID string, SALARY  int , COMMISSION_PCT string,MANAGER_ID string, DEPARTMENT_ID int , _rescued_data string',
    mode => 'PERMISSIVE',
    columnNameOfCorruptRecord => '_rescued_data'
))
```

```python
%sql
select * from employees
```

Records which has mismatch on schema, are loaded to _rescued_data column. In a row if some columns are matching the schema, they are loaded, along with mismatched full row in rescued column  - mode => 'PERMISSIVE'
