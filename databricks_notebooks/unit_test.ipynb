```python
from pyspark.sql.functions import *

def add_new_col(df, new, s_col):
    return (df.withColumn(new,                        
            when(col(s_col) == 0, 'Normal')                            
            .otherwise('Unknown')))
```

```python
from pyspark.testing.utils import assertDataFrameEqual

def test_add_new_col():
   data = [(0,), (1,), (-1,),(None,)]
   columns = ["value"]
   df = spark.createDataFrame(data, columns)

   actual_df = add_new_col(df, "new_value", "value")

   expected_data = [(0, 'Normal'), (1, 'Unknown'),(-1, 'Unknown'), (None, 'Unknown')]
   expected_df = spark.createDataFrame(expected_data,["value", "new_value"])

   assertDataFrameEqual(actual_df, expected_df)
test_add_new_col()
```
