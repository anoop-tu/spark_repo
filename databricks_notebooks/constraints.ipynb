```python
%sql
create or replace temp view incoming as 
select * from values
(8, 'SKU8', 'Product8', 'Category8', 60, 600.00),
(9, 'SKU9', 'Product9', 'Category9', 0, 700.00)
as t(order_id, sku, product_name, product_category, qty, unit_price);
```

```python
%sql
merge into orders1 as target
using incoming as source
on target.order_id = source.order_id
when matched then update set *
when not matched then insert *;
```
