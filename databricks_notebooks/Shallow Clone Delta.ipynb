```python
%sql
drop table if exists orders_managed_sclone 
```

Make a shallow clone

```python
%sql
create or replace table orders_managed_sclone shallow clone orders_managed;
describe detail  orders_managed_sclone;
```

Cloned table points to exact same parquet files as the source table

```python
%sql
select distinct _metadata.file_path from orders_managed_sclone
```

```python
%sql
select * from orders_managed_sclone
```

After cloning, even if we insert new data to source table, it wont reflect in cloned table . Its same vicer versa also

```python
%sql
insert into orders_managed (order_id, sku, product_name, product_category, qty, unit_price) values (101, '12354', 'product10', 'category10', 10, 100.00), (201, '12355', 'product20', 'category20', 20, 200.00)
```

```python
%sql
select * from orders_managed_sclone
```

```python
%sql
describe history orders_managed_sclone
```

```python
%sql
delete from orders_managed
```

```python
%sql
ALTER TABLE orders_managed
SET TBLPROPERTIES (
'delta.deletedFileRetentionDuration' = 'interval 0 hours'
)
```

```python
%sql
vacuum orders_managed
```

Even though source table is emptied, shallow clone remains intact

```python
%sql
select * from orders_managed_sclone
```

```python
%sql
drop table orders_managed
```

```python
%sql
select * from orders_managed_sclone
```

As long as the source files are available, shallow clone will show the data. Once files are deleted physically, shallow clone will fail to show any rows
