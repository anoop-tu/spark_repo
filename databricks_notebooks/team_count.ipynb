```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import *
from pyspark.sql import Window
spark = SparkSession. \
builder. \
appName("week5-assignment-2"). \
config('spark.ui.port','0'). \
config("spark.sql.warehouse.dir", f"/user/itv024128/warehouse"). \
enableHiveSupport(). \
master('yarn'). \
getOrCreate()
```

```python
data = [(1,8),(2,8),(3,8),(4,7),(5,9),(6,9)]
```

```python
schema = ["emp_id","team_id"]
```

```python
emp_df = spark.createDataFrame(data,schema)
```

```python
emp_df.show()
```

```python
window = Window.partitionBy("team_id")
```

```python
new_df = emp_df.withColumn("team_count", count("emp_id").over(window))
```
