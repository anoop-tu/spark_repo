```python
%sql
merge into ext_cat.default.cust1  target
using ext_cat.default.cust2  source
on target.customerid = source.customerid
when matched then update set *
when not matched then insert  *
when not matched by source then delete;
```

```python
from delta.tables import DeltaTable
from pyspark.sql import SparkSession

spark = SparkSession.builder.getOrCreate()

# Load source and target as Delta tables
target = DeltaTable.forName(spark, "ext_cat.default.cust1")
source = spark.table("ext_cat.default.cust2")

# Perform the merge (equivalent to SQL MERGE)
(
    target.merge(
        source,
        "target.customerid = source.customerid"  # ON condition
    )
    .whenMatchedUpdateAll()           # WHEN MATCHED THEN UPDATE SET *
    .whenNotMatchedInsertAll()        # WHEN NOT MATCHED THEN INSERT *
    .whenNotMatchedBySourceDelete()   # WHEN NOT MATCHED BY SOURCE THEN DELETE
    .execute()
)
```
