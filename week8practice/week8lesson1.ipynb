```python
from pyspark.sql import SparkSession

spark = SparkSession. \
builder. \
appName("week7-assignment-3"). \
config("spark.sql.warehouse.dir", f"/user/itv024128/warehouse"). \
enableHiveSupport(). \
master('yarn'). \
getOrCreate()
from pyspark.sql.types import *
from pyspark.sql.functions import *
```

```python
schema = 'booking_id long,guest_name string , checkin_date date , checkout_date date , room_type string , total_price double'
```

```python
hotel_df = spark.read.schema(schema).csv('/public/trendytech/datasets/hotel_data.csv')
```

```python
hotel_df.show(4)
```

### Variations of df.select()

```python
hotel_df.select("*").show(3)
```

```python
hotel_df.select("booking_id","total_price").show(3) #columns string notation
```

```python
hotel_df.select("booking_id",hotel_df.total_price,hotel_df['guest_name'],col('checkin_date')).show(3)
```

### expr()

```python
hotel_df.select(expr("checkout_date -  checkin_date")).show(3)
```

### col().like()

```python
hotel_df.select("booking_id",hotel_df.total_price,hotel_df['guest_name'],col('checkin_date')).where(col('guest_name').like('J%')).show(5)
```

```python
hotel_df.select("booking_id",hotel_df.total_price,hotel_df['guest_name'],col('checkin_date')).where('guest_name like "J%"').show(5)
```
